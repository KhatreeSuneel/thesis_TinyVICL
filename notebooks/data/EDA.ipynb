## Distribution of Dataset

```python
def plot_countplot(df, col, hue) -> None:
    plt.figure(figsize=(8, 4))
    plt.title(f"Distribution of column '{col}' $(n={len(df)})$", size=14)
    g = sns.countplot(data=df, x=col, edgecolor="black", hue=hue)
    for i in range(len(g.containers)):
        g.bar_label(g.containers[i])


plot_countplot(df, col="dataset", hue="task")
```

```python
def plot_newcountplot(df, col, hue) -> None:
    plt.figure(figsize=(10, 5))  
    
    plt.title(f"Distribution of column '{col}' $(n={len(df)})$", size=14)
    
    g = sns.countplot(data=df, x=col, edgecolor="black", hue=hue)
    
    
    for container in g.containers:
        g.bar_label(container, padding=3) 
    
    
    plt.ylim(0, plt.ylim()[1] * 1.15)
    
    plt.show()

plot_newcountplot(df, col="dataset", hue="task")
```
